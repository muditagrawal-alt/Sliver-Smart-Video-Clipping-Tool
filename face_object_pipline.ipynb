```python
# =========================
# ENVIRONMENT CHECK
# =========================
import sys
import os

print("Python executable:")
print(sys.executable)
print("Python version:")
print(sys.version)
print("-" * 50)

# =========================
# DEPENDENCY CHECK
# =========================
deps = {
    "cv2": "opencv-python",
    "torch": "torch",
    "ultralytics": "ultralytics",
    "numpy": "numpy",
    "ffmpeg": "ffmpeg-python",
    "deep_sort_realtime": "deep-sort-realtime",
}

failed = []

for module, name in deps.items():
    try:
        __import__(module)
        print(f"✅ {name} imported successfully")
    except Exception as e:
        print(f"❌ {name} FAILED: {e}")
        failed.append(name)

print("-" * 50)

if failed:
    raise RuntimeError(f"Dependency check failed for: {failed}")

print("All dependencies imported correctly ✅")
print("-" * 50)

# =========================
# MODEL FILE CHECK
# =========================
BASE_DIR = os.getcwd()
FACE_MODEL_PATH = "face_clip/models/yolov8n-face-lindevs.pt"
OBJ_MODEL_PATH = "face_clip/models/yolo11m.pt"

print("Working directory:")
print(BASE_DIR)
print("-" * 50)

for path in [FACE_MODEL_PATH, OBJ_MODEL_PATH]:
    if not os.path.exists(path):
        raise FileNotFoundError(f"❌ Model not found: {path}")
    else:
        size_mb = os.path.getsize(path) / (1024 * 1024)
        print(f"✅ Found {path} ({size_mb:.2f} MB)")

print("-" * 50)

# =========================
# MODEL LOAD CHECK
# =========================
from ultralytics import YOLO

print("Loading face model...")
face_model = YOLO(FACE_MODEL_PATH)
print("✅ Face model loaded")

print("Loading object model...")
object_model = YOLO(OBJ_MODEL_PATH)
print("✅ Object model loaded")

print("-" * 50)
print("🎯 ALL CHECKS PASSED — SYSTEM READY")
```
